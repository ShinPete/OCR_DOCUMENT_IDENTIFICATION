```python
# src/models/transformer_train.py (sketch)
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
import numpy as np, pandas as pd
from sklearn.metrics import f1_score, classification_report

model_id = "distilbert-base-uncased"
tok = AutoTokenizer.from_pretrained(model_id)

def to_ds(df, labels):
    enc = tok(df["text"].astype(str).tolist(), truncation=True, padding=True, max_length=256)
    enc["labels"] = [labels.index(y) for y in df["type"]]
    return Dataset.from_dict(enc)

train = pd.read_csv(REPO/"data/processed/datasets/v1/train.csv")
val   = pd.read_csv(REPO/"data/processed/datasets/v1/validation.csv")
labels = sorted(train["type"].unique())
ds_tr, ds_va = to_ds(train, labels), to_ds(val, labels)

model = AutoModelForSequenceClassification.from_pretrained(model_id, num_labels=len(labels))
args = TrainingArguments(
    output_dir="models/hf",
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=3,
    evaluation_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    save_total_limit=1,
    fp16=False,
    report_to="none",
)
def compute_metrics(p):
    preds = np.argmax(p.predictions, axis=1)
    f1 = f1_score(p.label_ids, preds, average="macro")
    return {"f1": f1}

trainer = Trainer(model=model, args=args, train_dataset=ds_tr, eval_dataset=ds_va, compute_metrics=compute_metrics)
trainer.train()
```

```python
def light_scrub(s: str) -> str:
    # drop quoted replies like "> On Tue..." but keep Subject and From
    import re
    return re.sub(r"(?m)^\s*>\s+.*$", "", s or "").strip()

def heavy_scrub(s: str) -> str:
    # remove common headers and signature-like tails
    import re
    t = re.sub(r"(?im)^(subject:.*$|from:.*$|to:.*$|date:.*$)", "", s or "")
    t = re.sub(r"(?m)^\s*>\s+.*$", "", t)
    t = re.sub(r"(?is)(\n--+\s*$|\nbest regards[!,.\s]*\n|\nthanks[!,.\s]*\n).*", "", t)
    return " ".join(t.split())

configs = [
  ("none", lambda x: x),
  ("light", light_scrub),
  ("heavy", heavy_scrub),
]
```

```python
train['light_scrub'] = train['text'].apply(light_scrub)
val['light_scrub'] = val['text'].apply(light_scrub)
train['heavy_scrub'] = train['text'].apply(heavy_scrub)
val['heavy_scrub'] = val['text'].apply(heavy_scrub)
```

```python

def run_cfg(analyzer, ngram_range, train, col,  val):

    pipe = Pipeline([
        ("tfidf", TfidfVectorizer(analyzer = analyzer, ngram_range=ngram_range, max_features=100000, min_df=2, sublinear_tf=True)),
        ("clf",   LogisticRegression(max_iter=2000, solver="lbfgs", n_jobs=None, class_weight="balanced", multi_class="auto"))
    ])
    pipe.fit(train[col].astype(str), train['type'])
    y_hat = pipe.predict(val[col])
    print(classification_report(val['type'], y_hat, zero_division=0))
    return pipe
```
